# Critical Exponents

The critical exponents, as outlined in the theory section, are another way to find $T_c$, and also generally help to describe the behaviour of different macroscopic properties nearby the critical temperature. 

Here, system sizes of 8-256 were used (8, 16, 32, 48, 64, 128, 256).

### Finding the maximum value of the susceptibility

Finding the maximum value of the suscpetibility was critical to finding the critical exponents, as well as the value of $T$ for which they occur. As shown in Fig. 1, the susceptibility (here shown as $\log\chi$) forms curves with certain peaks, which can be approximated by a polynomial. 

![Fig. 1: Collected susceptibility data.](/images/sus_plot_1.png)

However, even by limiting the range, it was found that plotting a simple polynomial around some "guessed" peak was inacurate over many runs (important for the forthcoming bootstrap error method), due to the noise in the data. The **Savitsky-Golay** approach was used to smooth the data, then find a maximum from this smoothed data (see Fig. 2). It is much more consistent due to the smoothing, as well as having little effect on important properties, including peak height (which is important here)

![Fig. 2: Savitsky-Golay fitting for the susceptibility data, with peaks labelled.](/images/savgol_1.png)


### Error on the maximum

The error can be estimated from the peak points by running a bootstrap method on the data. This essentially runs the fitting function multiple times, where the datapoints are skewed by a certain amount related to their error (specifically they are skewed by a normal distribution with standard deviation equal to their error). This means that error estimates come from the error on the original data points, which is important here. We also find error in the temperature values of the maximums.

It was important to use a long range of $T$ here, because the lower system sizes needed "time" to reach their maximum. Clearly, if we only went to $T=2.4$ then the maximum for the $8\times 8$ system simply wouldn't exist (in this range). 

### Finding the first critical exponent 

The first critical exponent to find is $\nu$, as it can be found independently from the other parameters. This is done by plotting the *psuedocritical point* of each system against its size. From the Theory section, this should follow a relationship:

$$
\bigg(T_c(N^2) - T_c(\infty)\bigg) \propto N^{-1/\nu}
$$

or simply 

$$
T_{ps} = T_c + AN^{-1/\nu}
$$

By using `scipy.optimize.curve_fit` and setting the fit function to the above equation, it can be found from this data that $\nu = 1.00 \pm 0.07$, equalling the analytical value, as can be seen in Fig. 3-4.

![Fig. 3: Pseudocritical temperature vs. system size, as expected in the theory. Fitted using $T_{ps}=T_c + AN^{-1/\nu}$ with values $T_c=2.2742 \pm 0.0006, $1/\nu = 1.00 \pm 0.07$, and $A=1.7 \pm 0.5$.](/images/ps_temp_vs_size_1.png)


![Fig. 4: Log-log plot of pseudocritical temperature versus system size, with values for the critical temperature, p and A from Fig 3 used to plot line of best fit and error.](/images/log_ps_temp_vs_size_1.png)

The fitting method has a large error due to fitting directly to the data, rather than directly to the log-log plot as done for the other critical exponenets. This was done because $T_c$ was assumed unknown before, and in order to observe a linear relationship in the log plot, this value must be subtraceted from each $T_{ps}$. Hence, fitting directly in Fig. 3. was necessary to determine this first, as well as the other factors - however as discussed this then leads to a large error due to the nature of this fit. Regardless, a value of $\nu = 1.000 \pm 0.064$ (error on inverse is $\Delta \nu = \nu^2 \Delta (1/\nu)$) is very close to the expected value of $1$ from the analytical solution.  

### Finding the second critical exponent 

The second critical exponent to find is $\gamma$, which can be computed from a log-log plot of $\chi_\text{max}$ aginst system size (as seen in the Theory section). The slope of this plot is equal to $\gamma / \nu$, and since we know $\nu=1.00 \pm 0.07$, is just equal to $\gamma$ (plus the error on $\mu$). This can be seen in Fig. 5-6. 

![Fig. 5: A plot of the maximum value of the susceptibility against system size.](/images/max_chi_vs_size_1.png)

![Fig. 6: Log-log plot of maximum susceptibility versus system size, fitted using a linear fit. Values of m and c from Fig. 5 are used to plot line of best fit and error. This finds a slope of $1.758 \pm 0.005$. The system of size $256 \times 256$ is excluded from the data here, as it does not follow the linear fit accurately.](/images/log_max_chi_vs_size_1.png)

This finds a slighlty unexpected value of $\gamma = 1.76 \pm 0.12$ (error calculated via $\Delta \gamma = \gamma \sqrt{(\frac{\Delta(\gamma/\nu)}{\gamma/\nu})^2 + (\frac{\Delta (1/\nu)}{1/\nu})^2}$). This is reasonable close to the expected value of 1.75, however it does have a large error, resulting from the large error on $\mu$, which propogates into the error on $\gamma$. 

### Finding the third critical exponent

The third critical exponent to find is $\beta$. This is found in a very similar way to the previous two critical exponents. 

![Fig. 7: A log-log plot of the root-mean-square magnetisation versus system size, fitted using a linear fit.](/images/loglog_Mrms_vs_size.png)

This finds a value of $\beta=0.1268\pm 0.0089$ (using the same error formula as before). This is not far from the expected value of 0.125. 



### Finding the fourth critical exponent

The fourth and final critical exponent to find is $\alpha$. This was notably more difficult to compute than the other three critical exponents, namely because it is smaller than all of them (analytically $alpha=0$). To find it, we follow the same approach as with $\gamma$, except now plotting the mamximum of the specific heat of each system, as can be seen in Fig. 10.

![Fig. 10: A plot of $\log C_V$ vs. $T$ for the different system sizes. Notice immediately how the smaller systems diverge from $T_c$ massively. This can also been seen for larger and larger systems. Sizes 128 and 256 are close together, and with larger systems, one would expect them to converge to the same point on the graph (computational limitations meant that these large systems could not be computed in a reasonable amount of time).](/images/log_cv_vs_T.png)

Removing the smaller sizes (8, 16) due to their divergence from the expected result, and running bootstrap to find errors as before, a graph of $\log C_V$ against $\log N$ can be plotted, with the gradient being $\alpha/\nu$, as can be seen in Fig. 11.

![Fig. 11: A plot of $\log C_V$ vs. $\log N$. This produces a gradient of $\alpha/\nu = 0.1447 \pm 0.0085$. An important note is that these systems do not produce a "flat" linear fit as we would expect from the analytical value of $\alpha$.](/images/log_cv_vs_log_N.png)

Using the gradient and the value of $\nu$ found earlier, it can be deduced that $\alpha=0.145 \pm 0.013$, which is a relatively far deviation from the expected value of $\alpha=0$. However, as discussed, this is likely due to using small system sizes, which diverge from the expected values of $C_V$ at the critical temperature. Looking at Fig. 11, it might not be too much of a stretch to say that it seems to be "flattening" out for larger systems, and so implementing more large systems into consideration would potentially drive $\alpha$ closer to its expected value.

### Discussion

The crtical exponents here align with the expected values found analytically quite well. However, there is a large error due to the fitting done to originally find $\nu$, which transpires into a large error on other critical exponents. 

| Critical Exponent Found | Value | Error |
| ----------------------- | ----- | ----- |
| $\nu$                   | 1.000 | 0.064 |
| $\gamma$                | 1.76  | 0.12  | 
| $\beta$                 | 0.1268 | 0.0089 | 
| $\alpha$                | 0.145 | 0.013 |

The large error in the fitting for $\nu$ ultimately arises from a large error in the maximum susceptibility for the smaller system sizes, since they have much "flatter" distributions, and are more likely to have large shifts in a maximum. The smoothing function did help to avoid this somewhat and reduce the error, however it is still large. Removing these from the fitting also is not successful, since then the fit has too few data points to fit to and provides wildy innacurate results. A better approach then might be to use a finer range of system sizes (with more system sizes overall), however due to computational limitations this was deemed not plausible. 

Running the system for longer may also have stabilised the smaller systems, however, for the same reason stated above, this was deemed not plausible.
