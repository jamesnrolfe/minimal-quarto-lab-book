# Optimisations

A few optimisations were introduced to improve the speed of the code, allowing us to perform operations on larger system sizes in much shorter timeframes. 

Firstly, the Python library **Numba** was implemented to some functions, which drastically increased the speed at which the program executed. Numba is a *just-in-time compiler*, which compiles certain functions directly machine code, making them faster at runtime. Python is somewhat an *interpreted* language, which means that a line is "translated" into machine code as it is run. However, Python is actually compiled into *bytecode* before running, which is a simplified version similar to assembly; this is done to increase the interpretation speed. Regardless, compiling direct to machine code reduces the workload of the interpreter, and hence the function runs much faster.

Numba doesn't work particularly well with certain libraries and functions within Python, however it does work with **Numpy**, which is the primary maths module used here. Numpy is already fast at certain tasks (most parts that require fast computation are written in **C** or **C++**), and so this in combination with Numba should produce excellent results. 

In fact, upon running functions with the Numbra wrapper enabled (`mcmove` as an example, the function to run the Monte Carlo evolution), evolving the system runs approximately 10 times faster - a massive speed increase. 

A few other quality-of-life optimisations were made. The code was refactored into separate Python files, and then imported into the main Jupyiter notebook for ease of access. New modules listed below were added to improve usability. 

1. **tqdm** provides loading bars for iteration in Python. This provides valuable information about roughly how long a loop will take to run, and hence time is not wasted waiting for a "random" amount of time.

2. **reload** automatically refreshes imported files, which is critical here when using a Jupyiter notebook.

As well as this, the code was rewritten to include a more consistent and readable variable name structure, to improve its understandability and readability.
