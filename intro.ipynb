# Introduction

The **Ising model** (or **Lenz-Ising model**), first proposed by Wilhelm Lenz in the early 1920s [1], is a theoretical model used to describe magnetism. The model is developed on the concept of *spin*, in which each point on a lattice can have spin "up" or "down". These spins favour alignment with their closest neighbours, and so the Ising model has been successful in explaining the emergence of **ferromagnetism**. Furthermore, the Ising model is relevant to the study of **phase transitions**, the processes through which a material changes is macroscopic state. 

The Ising model offers essential insight into **critical behaviour**, describing how systems behave close to critical points, and **universailty**, where different systems appear to demonstrate similar properties close to critical points. 

A student of Lenz, Ernst Ising successfully analytically solved the Ising model in one-dimension in 1925 [2], finding that in this case, assuming short-range interactions, there are no phase transitions. The two-dimension case also has a analytical solution for square lattices with no applied external magnetic field, found by Lars Onsager in 1944 [3]. Onsager discovered that a nonzero "critical temperature" $T_c$ is found - below which thermal fluctuations are insufficient to destroy spontaneous magnetisation (i.e. ferromagnetism). Systems have interesting properties at the critical temperature, often displaying *scale-invariance*. This critical temperature is key to statistical physics. 

Despite not having analytical solutions for higher dimensions, the Ising model can be explored numerically using methods such as Monte-Carlo simulations, where these critical phenomena can be observed.
