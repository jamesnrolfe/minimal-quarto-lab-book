# Introduction

The *Ising model* (or *Lenz*-Ising model) is a theoretical model often used to describe Magnetism, which is often best utilised numerically (in fact, it is currently impossible to solve generally). It helps to explain the effect of *ferromagnetism*, wherby magnetic moments align despite the removal of an applied magnetic field.  

It also provides great insight into *phase transitions*, whereby a material changes state (such as water boiling in a kettle to become steam). 

The Ising model has had many other uses, such as uses statistically modelling neuron activity[1] and hence a natural extension has been found with neural networks, as well as to model the motion of atoms in gases.
