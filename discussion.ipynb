# Discussion

Overall, the experiment produced some good, accurate results, summarised here for convenience.

**Critical Temperature via Binder Cumulant results**

| Method                            | Result                | $\%$ error from analytical value |
| --------------------------------- | --------------------- | -------------------------------- |
| Inv. Var. bootstrap (large range) | $2.256 \pm 0.057$     | $0.58\%$                         |
| Inv. Var. bootstrap (low range)   | $2.26878 \pm 0.00058$ | $0.018\%$                        |
| Weighted average pairwise         | $2.26894 \pm 0.00015$ | $0.011\%$                        |
| Extrapolating pairwise temps      | $2.26911 \pm 0.00024$ | $0.0032\%$                       |

**Critical exponent Results**

| Critical Exponent Found | Value | Error | $\%$ error from analyitcal value |
| ----------------------- | ----- | ----- | -------------------------------- |
| $\nu$                   | 1.000 | 0.064 | $0\%$ |
| $\gamma$                | 1.76  | 0.12  | $0.57\%$ |
| $\beta$                 | 0.1268 | 0.0089 | $1.44\%$ | 
| $\alpha$                | 0.145 | 0.013 | N/A |

Considering the critical temperature extrapolated from the pairwise crossings, it is clear that this result is in line with the theory (within $0.0032\%$ of the expected analytic value). This confirms that the Binder-cumulant crossing approach is reliable in pinpointing $T_c$ in the 2D Ising Model (with a square lattice). 

This finite-size scaling analysis presented here illustrates several key features of critical phenomena. Firstly, the Binder cumulants of the systems are observed to intersect at an approximately common point (see Fig. \ref{fig: spline_fit}). Combining the Swendsen-Wang algorithm with corrections for finite-size effects (as recommended by Binder and Heermann \cite{Binder_Heerman_2019}, see Fig. \ref{fig: extrap}), this provides a robust estimate for the critical temperature $T_c=(2.26911 \pm 0.00024) \frac{J}{k_B}$. This result is within $0.0032\%$ of the analytical value. Fitting to raw data prior to Fig. \ref{fig:critexp_plots}.a provided another estimate for $T_c\approx 2.274\frac{J}{k_B}$, however this value differs from the true value by $\sim 0.2\%$ due to larger finite-size effects. 

By measuring and analysing certain macroscopic properties of each system, and how these properties scale with system size (see Fig. \ref{fig:critexp_plots}), the critical exponents $\nu$, $\gamma$, $\alpha$ and $\beta$ could be extracted (see Table \ref{tab: crit_exp_results}).  $\nu$ is close to its critical value, but with a large error of $6.4\%$. The critical temperature was not assumed to be known a priori, and so fitting linearly on a log-plot was not possible; instead fitting a exponential function to the raw data. This procedure introduced a large error on the fit parameter $\nu$, which propagates through to the other critical exponents, all of which depend on $\nu$. An improvement here would be to use a finer set of system sizes to reduce the error on this fit, however, due to computational limitations, this was not deemed feasible. As well as this, the data for the specific heat in Fig. \ref{fig:critexp_plots}.d deviates noticeably from the linear trend expected from a power-law fit. Since theoretically $\alpha=0$, it is implied that there is a weak or logarithmic divergence near specific heat. As a result, the data points are dominated by sub-leading corrections rather than the main power-law behaviour. A logarithmic fit to the raw data would provide a more stable means of capturing the behaviour near criticality. 

Generally, improvements to the data could be made with longer simulations, allowing the systems to explore more of their phase spaces, especially near criticality. Computational limitations here thus increased the uncertainty in the results. 
    
Data from the Ising model has interesting and far-reaching applications from helping to model superconductivity in magnetic monolayers [1] to simulating social dynamics [2]. Comparing the data found here to experimental data on physical materials would provide robust support to the universality of the Ising models predictions.
